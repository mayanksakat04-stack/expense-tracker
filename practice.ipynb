#### Problem 1: Find the largest number without using max()

```python
def find_largenum(numbers):
    max_num = numbers[0]
    for i in range(len(numbers)):
        if numbers[i] > max_num:
            max_num = numbers[i]
    return max_num

numbers = [12, 45, 2, 89, 34, 100]
print(find_largenum(numbers))
```

#### Problem 2: Count how many vowels are present.

```python
def count_vowels(word):
    counter = 0
    for letter in word:
        if letter in ['a','e','i','o','u']:
            counter += 1
    return counter

word = "education"
print(count_vowels(word))
```
